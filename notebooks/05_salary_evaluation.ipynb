# Phase 5 — Salary Evaluation

Run `scripts/salary_tiers.py` to:
1. Compute within-archetype percentile ranks (WS/48, BPM, VORP)
2. Assign Performance Tier (T1–T5) per player
3. Assign Salary Tier (T1–T5) per player
4. Compute Delta_Tier and verdict (FAIR VALUE / OVERPAID / UNDERPAID)
5. Generate player card narratives via `scripts/llm_narrative.py`

Outputs: `data/nba_salary_evaluation.csv`, `reports/player_cards/`

See `plan.md §5 Phase 5` and `plan.md §2 Layer 6` for the tier definitions.
