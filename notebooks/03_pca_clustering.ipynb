# Phase 3 — PCA + Archetype Clustering

Run `scripts/clustering.py` to:
1. Z-score normalize Tool scores
2. Run PCA (>= 80% variance threshold)
3. Run K-Means (k=6–10), select k by silhouette score
4. Label clusters via player_type cosine matching; LLM for residuals

Outputs: `data/nba_archetypes.csv`, `reports/cluster_profiles.json`

See `plan.md §5 Phase 3` and `workflow.plan.md §2` for details.
