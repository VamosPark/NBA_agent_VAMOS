# Phase 4 — Similarity + Synergy Analysis

Run `scripts/similarity.py` to:
1. Compute cosine similarity matrix on PCA scores → top 10 similar players per player-season
2. Compute interaction-effect synergy for pairs with >= 100 shared minutes
3. Assign team_context_tier (quintile of team baseline net rating)
4. Fallback to archetype complementarity for thin-sample pairs

Outputs: `data/player_similarity.csv`, `data/lineup_synergy.csv`

See `plan.md §5 Phase 4` and `workflow.plan.md §3` for the synergy formula.
